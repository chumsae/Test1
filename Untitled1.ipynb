```python
#연산자와 조건문 과제

age = 15

if age >= 20:
  print("성인입니다.")
elif age >= 14:
  print("청소년입니다.")
elif age >= 8:
  print("초등학생입니다.")
```

```python
#연산자와 조건문 과제

height_cm = 163.2
weight_kg = 64
bmi = weight_kg / ((height_cm / 100) ** 2)

if bmi >= 25:
  print("과체중입니다")
else:
  print("정상 체중입니다")
```

```python
#논리 연산자와 조건문 과제

is_weekend = True
has_homework = False

if is_weekend and not has_homework:
  print("푹 쉴 수 있어요")
elif is_weekend and has_homework:
  print("숙제를 먼저 해요")
elif not is_weekend and has_homework:
  print("오늘도 바빠요")
else:
  print("조금 여유롭네요")

```

```python
# 위 리스트에서 이름을 하나씩 꺼내서 "안녕하세요, ○○님" 형식으로 출력해보세요.

names = ["지민", "수아", "하늘", "민재"]

for name in names:
    print("안녕하세요,", name, "님")
```

```python
# 힌트: 변수를 하나 만들고 3보다 작을 동안 출력하고, 1씩 증가시켜보세요.
i = 1


while i <= 3:
    print("파이썬 재밌어요")

    i = i + 1
```

```python
# 힌트: 시작값은 10, 조건은 1보다 크거나 같을 때, 매번 -1씩 줄이기

i = 10
while i >= 1:
    print(i)

    i = i - 1
```
