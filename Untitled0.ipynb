```python
#연산자와 조건문 과제

height_cm = 163.2
weight_kg = 64
bmi = weight_kg / ((height_cm / 100) ** 2)

if bmi >= 25:
  print("과체중입니다")
else:
  print("정상 체중입니다")
```

```python
#논리 연산자와 조건문 과제

is_weekend = True
has_homework = False

if is_weekend and not has_homework:
  print("푹 쉴 수 있어요")
elif is_weekend and has_homework:
  print("숙제를 먼저 해요")
elif not is_weekend and has_homework:
  print("오늘도 바빠요")
else:
  print("조금 여유롭네요")

```
